```python
## pacing  tuple
packed_tuple =1,'hello',3.15
print(packed_tuple)
```

```python
## unpacing a tuple
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## unpacking with *
numbers = (1,2,3,4,5)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## nested tuple
# nested list
lst = [[1,2,3,4],[6,7,8,9],[1,'hello',3.14,c]]
print(lst[0])
print(lst[2])
print(lst[0][0:3])
```

```python
lst = [[1,2,3,4],[6,7,8,9],(1,'hello',3.14,c)]
print(lst[2][0:3])
```

```python
nested_tuple = ((1,2,3),('a','b','c'),(True,False))
## access the elements inside the tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuple
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
```
