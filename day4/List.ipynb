## list comprehension
Basic syntax [expression for in item in iterables]
with condition logic [epresssion for item in iterable in condition]
nested list comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
## basic list comprehension
square = [num**2 for num in range(10)]
print(square)
```

```python
### comprehension with condition
list = []
for i in range(10):
    if i%2==0:
        list.append(i)
print(list)
```

```python
## list comprehention
even =[num for num in range(10) if num%2==0]
print(even)
```

```python
## nested list comprehension
list1 = [1,2,3,4]
list2 = ['a','b','c','d']

pair = [(i,j)for i in list1 for j in list2]
print(pair)
```

```python
## list comprehension with fnction call (inbuild)
words = ['hello','world','python','list','comprehension']
lengths = [len(word) for word in words]
print(lengths)
```
